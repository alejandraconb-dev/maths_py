```python
import sympy
from sympy import symbols, exp, sqrt, diff

x = symbols('x')
y = exp(sqrt(x))

dy_dx = diff(y, x)

print(f"The derivative of y = e^(x^(1/2)) is: {dy_dx}")
```

```python
import sympy
from sympy import symbols, exp, sqrt, diff

x = symbols('x')
y = 1 / (3 - x**3)

dy_dx = diff(y, x)

print(f"The derivative of y = 1 / (3 - x^3) is: {dy_dx}")
```

```python
import sympy as sp

# 1. Definir la variable simbólica
x = sp.symbols('x')

# 2. Definir la función
# e^(x**(1/2)) es lo mismo que e^sqrt(x)
f_x = sp.exp(x**(sp.Rational(1, 2)))

# 3. Calcular la derivada
# sp.diff(función, variable)
derivada_f_x = sp.diff(f_x, x)

# 4. Mostrar el resultado
print("Función original:")
sp.pretty_print(f_x)

print("\nDerivada (dy/dx):")
sp.pretty_print(derivada_f_x)
```

```python
import sympy as sp

# 1. Definir la variable simbólica
x = sp.symbols('x')

# 2. Definir la función
# Usamos la sintaxis de Python para representar la función: 1 / (3 - x**3)
f_x = 1 / (3 - x**3)

# 3. Calcular la derivada
# sp.diff(función, variable)
derivada_f_x = sp.diff(f_x, x)

# 4. Mostrar el resultado
print("Función original:")
sp.pretty_print(f_x) # Muestra la función en un formato legible

print("\nDerivada (dy/dx):")
sp.pretty_print(derivada_f_x) # Muestra la derivada en un formato legible
```
